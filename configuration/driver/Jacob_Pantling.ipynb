```python
import numpy as np
import matplotlib.pyplot as plt

# Material properties
lambda_i = 2.2       # Thermal conductivity (W/m/K)
rho_i = 930          # Ice density (kg/m³)
L = 334000           # Latent heat of fusion (J/kg)

# Environmental conditions
Tf = 0            # Freezing temperature (°C)
Ta = -10             # Air temperature (°C)
ha = 20              # Heat transfer coefficient (W/m²/K)

def natural_growth(t):
    b_ = lambda_i / ha
    ac_ = 2 * lambda_i * (Tf - Ta) * t / (rho_i * L)

    eta = np.sqrt(b_**2 + ac_) - b_
    return eta

# Simulate 30 days
hours = np.linspace(0, 720, 500)
time_seconds = hours * 3600

thickness = natural_growth(time_seconds)

print(f"Ice growth after 30 days: {thickness[-1]*100:.2f} cm")

# Plot
plt.figure(figsize=(10, 6))
plt.plot(hours, thickness * 100, label="Jacob's analytical model FRESH WATER")

plt.xlabel("Time (hours)")
plt.ylabel("Additional basal ice thickness (cm)")
plt.title("Natural Ice Growth")
plt.grid(True)
plt.legend()
plt.show()
 
```
